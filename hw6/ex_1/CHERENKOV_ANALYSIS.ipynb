```python
ind_out_array = np.linspace(0,965,966)
#ind_out_array = np.linspace(300,302,2)

for step in ind_out_array:
    ex_array = np.array(fdata.isel(t=ind_out).Ex.load())
    ey_array = np.array(fdata.isel(t=ind_out).Ey.load())
    ez_array = np.array(fdata.isel(t=ind_out).Ez.load())
    x_array = np.array(fdata.isel(t=ind_out).x.load())
    y_array = np.array(fdata.isel(t=ind_out).x.load())
    
    # Example of a heat map of the electric energy E^
    emin = 0
    emax = 0.5

    # Compu2te field magnitude
    emag = np.sqrt(ex_array**2 + ey_array**2 + ez_array**2)

    plt.figure(figsize=(7, 7))
    plt.pcolormesh(
        x_array,
        y_array,
        emag,
        cmap='viridis',
        shading="auto",
        vmin=emin, 
        vmax=emax
    )
    plt.colorbar(label=r"Electric Field $|E|$", fraction=0.0465, pad=0.02)
    plt.gca().set_aspect("equal", adjustable="box")
    plt.title(r"Cherenkov Instability ($u_x = u_y \approx 70.71$):    step={}".format(step))
    plt.xlabel("x")
    plt.ylabel("y")
    plt.tight_layout()
    plt.savefig('Efield_70/step_{}.png'.format(step))
    plt.close()
```

D) Compare the growth of the total electric field in the domain for both cases by calculating the sum of $E^2$ at each time step. 
- Plot the time evolution of the electric energy for both cases in the same figure.
- How do the evolutions compare, and which streaming direction leads to faster growth of the numerical instability?
- Do you observe a clear difference in growth rates or onset times?

```python
# Load the simulation data from disk
# 'Data' is the main container for the simulation dataset (replace with your location)
data = nt2.Data("/dartfs/rc/lab/E/EPaCO/phys118/godfrey/hw6/run1/cherenkov_boris_vxy")
```

```python
# Select field data 
fdata = data.fields
```

```python
# Example loop for extracting the Electric energy from all recorded timesteps (adapt range as needed)
eenrg = []
for ind_out in tqdm(range(0, 966)):

    ex_array = np.array(fdata.isel(t=ind_out).Ex.load())
    ey_array = np.array(fdata.isel(t=ind_out).Ey.load())
    ez_array = np.array(fdata.isel(t=ind_out).Ez.load())
    eenrg.append(np.sum((ex_array**2 + ey_array**2 + ez_array**2)))

eenrg = np.array(eenrg)
```

```python
# Load the simulation data from disk
# 'Data' is the main container for the simulation dataset (replace with your location)
data2 = nt2.Data("/dartfs/rc/lab/E/EPaCO/phys118/godfrey/hw6/run2/cherenkov_boris_vxy")
```

```python
# Select field data 
fdata2 = data2.fields
```

```python
# Example loop for extracting the Electric energy from all recorded timesteps (adapt range as needed)
eenrg_100 = []
for ind_out in tqdm(range(0, 966)):

    ex_array2 = np.array(fdata2.isel(t=ind_out).Ex.load())
    ey_array2 = np.array(fdata2.isel(t=ind_out).Ey.load())
    ez_array2 = np.array(fdata2.isel(t=ind_out).Ez.load())
    eenrg_100.append(np.sum((ex_array2**2 + ey_array2**2 + ez_array2**2)))

eenrg_100 = np.array(eenrg_100)
```

```python
# BOTH ON SAME PLOT

t = np.linspace(0,965,966)
plt.plot(t, eenrg, label=r'$u_x = u_y \approx 70.71$')
plt.plot(t, eenrg_100, label=r'u=100')
plt.legend()
plt.title('Time Evolution of Electric Energy')
plt.xlabel('Time Step')
plt.ylabel(r'$\sum E^2$')
plt.tight_layout()
plt.savefig('time_evolution_both.png')
```
